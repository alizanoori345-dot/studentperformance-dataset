###DATA VISUALIZATION


```python
import matplotlib.pyplot as plt
#Line plotting->#line plot trends or time show karta hai
plt.plot(df['Study_Hours'],df['Final_Marks'])
plt.title('Study Hour vs Final Marks')
plt.xlabel('Study Hour')
plt.ylabel('Final Marks')
plt.show()
```

```python
#Bar Chart->used to compare different categories
plt.bar(df['Average_Marks'],df['Attendance'])
plt.xlabel("Average Marks")
plt.ylabel("Attendence")
plt.title("Attendence Vs Averge Marks")
plt.show()
```

```python
#Histogram->distribution of a dataset
plt.hist(df['Final_Marks'],bins=7)
plt.xlabel("Final Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Final Marks")
plt.show()
```

```python
#Scatter Plot-> shows the relationship between two variables.
plt.scatter(df['Study_Hours'],df['Final_Marks'])
plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("Study Hours vs Final Marks")
plt.show()
```

```python
#Heatmap->shows the relationship or intensity between variables using colors
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True )
plt.show()

```

```python
#Hypothesis Test (P-Value)
from scipy import stats
import numpy as np
data1=df['Study_Hours']
data2=df['Final_Marks']
t_stat, p_value = stats.ttest_rel(data1, data2)

print("t-statistic:",t_stat)
print("p-value:",p_value)
if p_value < 0.05:
  print("Rejected all null hypothesis: difference exit:")
else:
  print("Fail to reject null hypothesis: No significant difference")
```

```python
#Probability
passedstudents=(df['Final_Marks']>=50).sum()
total=len(df)
prob=passedstudents/total
print("Probability of Passing",prob)
```
